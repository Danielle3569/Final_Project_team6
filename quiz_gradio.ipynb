```python
def check_answer(user_answer):
    if list_answer[0][14] == user_answer:
        return "맞았습니다.\n\n"+list_answer[0]
    else: 
        return "틀렸습니다\n\n"+list_answer[0]
```

```python
with gr.Blocks() as quiz_page:
    with gr.Row():
        user_answer = gr.Radio(["A","B","C","D"],label='문제', info=list_quiz[0][3:])
        send_button = gr.Button('submit')
        text_box = gr.Textbox()
        send_button.click(fn=check_answer,inputs=user_answer,outputs=text_box)
```
